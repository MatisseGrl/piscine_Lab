```python
%pip install tokenizers==0.23.3

from pathlib import Path
import subprocess
import sys

# Repère les fichiers du mélange dans les inputs Kaggle.
dataset = next(
    path.parent for path in Path("/kaggle/input").rglob("manifest.json")
    if (path.parent / "equal" / "train.bin").is_file()
)
kit = Path("/kaggle/working/piscine-kit")
output = Path("/kaggle/working/runs/source-mix-equal")

# Récupère la même version du kit que pour le run de référence.
if not kit.exists():
    subprocess.run(
        ["git", "clone", "https://github.com/Meuleur/Piscine.git", str(kit)],
        check=True,
    )
subprocess.run(
    ["git", "checkout", "--detach", "3709c85ddaabfcd688d6eea80a5c5ed90605b549"],
    cwd=kit, check=True,
)

# Entraîne le modèle, puis mesure le BPB sur la validation publique.
subprocess.run([
    sys.executable, "-m", "ftb.train",
    "--data", str(dataset / "equal"),
    "--tokenizer", str(dataset / "tokenizer.json"),
    "--out", str(output), "--device", "cuda:0", "--micro-batch", "8",
], cwd=kit, check=True)

subprocess.run([
    sys.executable, "-m", "ftb.evaluate",
    "--ckpt", str(output / "ckpt.pt"),
    "--tokenizer", str(dataset / "tokenizer.json"),
    "--data", "data/validation.jsonl",
    "--device", "cuda:0", "--out", str(output / "evaluation.json"),
], cwd=kit, check=True)

print((output / "evaluation.json").read_text())
```
