```python
%pip install tokenizers==0.23.3

from pathlib import Path
import subprocess
import sys

# Repère le mélange 74/18/8 dans les fichiers ajoutés au notebook.
variant = "mix_74_18_8"
dataset = next(
    (
        path.parent for path in Path("/kaggle/input").rglob("manifest.json")
        if (path.parent / variant / "train.bin").is_file()
    ),
    None,
)
if dataset is None:
    raise FileNotFoundError("Ajoute source-mix-74-18-8.zip via Upload, puis relance.")

kit = Path("/kaggle/working/piscine-kit")
output = Path("/kaggle/working/runs/source-mix-74-18-8")
tokenizer = dataset / "tokenizer.json"

# Utilise la même version du kit que pour le mélange à parts égales.
if not kit.exists():
    subprocess.run(
        ["git", "clone", "https://github.com/Meuleur/Piscine.git", str(kit)],
        check=True,
    )
subprocess.run(
    ["git", "checkout", "--detach", "3709c85ddaabfcd688d6eea80a5c5ed90605b549"],
    cwd=kit,
    check=True,
)

# Entraîne le modèle avec le mélange 74/18/8.
subprocess.run([
    sys.executable, "-m", "ftb.train",
    "--data", str(dataset / variant),
    "--tokenizer", str(tokenizer),
    "--out", str(output),
    "--device", "cuda:0",
    "--micro-batch", "8",
], cwd=kit, check=True)

# Mesure le BPB final sur les 230 documents de validation.
evaluation = output / "evaluation.json"
subprocess.run([
    sys.executable, "-m", "ftb.evaluate",
    "--ckpt", str(output / "ckpt.pt"),
    "--tokenizer", str(tokenizer),
    "--data", "data/validation.jsonl",
    "--device", "cuda:0",
    "--out", str(evaluation),
], cwd=kit, check=True)

print(evaluation.read_text(encoding="utf-8"))
```
